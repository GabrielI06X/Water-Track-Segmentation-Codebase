```python
import os, random, uuid

img_dir = fr"/home/gaimholte/Imholte_segmentation_gym_ubuntu/segmentation_gym-main/High_Res_WT_Data/WT_Dataset_3D/WT_Data/Images_and_Labels/images"
msk_dir = fr"/home/gaimholte/Imholte_segmentation_gym_ubuntu/segmentation_gym-main/High_Res_WT_Data/WT_Dataset_3D/WT_Data/Images_and_Labels/labels"
ndvi_dir = fr"/home/gaimholte/Imholte_segmentation_gym_ubuntu/segmentation_gym-main/High_Res_WT_Data/WT_Dataset_3D/WT_Data/Images_and_Labels/ndvi"
slope_dir = fr'/home/gaimholte/Imholte_segmentation_gym_ubuntu/segmentation_gym-main/High_Res_WT_Data/WT_Dataset_3D/WT_Data/Images_and_Labels/slope'

# Zip and sort to keep them paired
imgs = sorted(os.listdir(img_dir))
msks = sorted(os.listdir(msk_dir))
ndvi = sorted(os.listdir(ndvi_dir))
slope = sorted(os.listdir(slope_dir))


for img, msk, ndvi, slope in zip(imgs, msks, ndvi, slope):
    # Generate a matching random prefix for both
    rand_prefix = str(uuid.uuid4())[:8] 
    
    os.rename(os.path.join(img_dir, img), os.path.join(img_dir, f"{rand_prefix}_{img}"))
    os.rename(os.path.join(msk_dir, msk), os.path.join(msk_dir, f"{rand_prefix}_{msk}"))
    os.rename(os.path.join(ndvi_dir, ndvi), os.path.join(ndvi_dir, f"{rand_prefix}_{ndvi}"))
    os.rename(os.path.join(slope_dir, slope), os.path.join(slope_dir, f"{rand_prefix}_{slope}"))
    #This block of code above references the first directory in the first argument, then creates the adjusted one within the second and renames the first one to that one. os.path.join() just creates a new directory by combining the filename and the main directory

```
