# **Purpose: RGBA Tile Filter**

Assesses RGBA tiles created by gdal_retile in QGIS in their directory and saves the ones with no missing data, as assessed by the alpha band, and saves them to a new folder in this notebook's directory named 'Usable_Images'.

```python
#Must change directory between uses
#Make sure the folder directory has no forward slash after it and is absolute; filename can be obtained by looping over that same directory with os.listdir

#Change this to the absolute directory of your tile folder that includes all tiles produced by retile
tile_folder_path = fr'C:\Users\gaimholte\OneDrive - William & Mary\Desktop\High_Res_WT_Images_and_Masks\Images\Masks_and_Sections_WV02_20190820222750_10300100968FB300_19AUG20222750-M1BS-503550221040_01_P004_u16rf3413\Sections_Color'

#Takes a filename with no accompanying directory, an empty bad list, an empty good list, and a folder directory in which each 'filename' resides.
#Places tiles with missing data into bad list and prints it is no good, while files with complete data are placed into the good list.
def find_no_data(filename, bad_list, good_list, folder_directory):
    image_alpha = rasterio.open(fr'{folder_directory}\{filename}').read(4)
    for row in image_alpha:
        for pixel in row:
            if pixel==0:
                print(fr'{filename} is no good!')
                bad_list.append(filename)
                return
    good_list.append(filename)


#Places good images into a folder within this notebook's directory
def save_good_files(good_list, folder_directory):
    new_path = Path('Usable_Images') #Can change folder name here
    new_path.mkdir(exist_ok=True)
    destination_path = new_path.resolve()
    for item in good_list:
        source_path = Path(fr'{folder_directory}/{item}')
        shutil.copy(source_path, destination_path)

```

```python
#Runs the script for RGBA tiles
bad = []
good = []
for item in os.listdir(tile_folder_path):
    find_no_data(item, bad, good, tile_folder_path)

save_good_files(good, tile_folder_path)
```

# **Purpose: NDVI Tile Filter**

Assuming single band imagery displaying a calculated NDVI, detects if any pixel values exceed 1 or are below -1, which indicates missing data after using gdal_retile, and saves images with complete data to a folder named 'Usable_NDVI_Images' within this notebook's directory

```python
#Change this to the absolute directory of your tile folder that includes all tiles produced by retile
ndvi_tile_folder_path = fr'C:\Users\gaimholte\OneDrive - William & Mary\Desktop\High_Res_WT_Images_and_Masks\Images\Full_Images_Masks_and_Sections_WV02_20190820222750_10300100968FB300_19AUG20222750-M1BS-503550221040_01_P004_u16rf3413\Sections_NDVI_Total'

#Counts pixels in an image; use for troubleshooting
def count_pixels_ndvi(filename):
    pixel_count=[]
    image_ndvi = rasterio.open(fr'D:\Imholte_Research_WT\High_Res_WT_Images_and_Masks\Notebooks\Retile_Missing_Data_Filter\{filename}').read(1)
    for row in image_ndvi:
        for pixel in row:
            pixel_count.append(pixel)
    return len(pixel_count)

#Takes a filename with no accompanying directory, an empty bad list, an empty good list, and a folder directory in which each 'filename' resides.
#Places tiles with missing data into bad list and prints it is no good, while files with complete data are placed into the good list.
def find_missing_data_ndvi(filename, bad_list, good_list, folder_directory):
    pixel_count=[]
    image_ndvi = rasterio.open(fr'{folder_directory}\{filename}').read(1)
    for row in image_ndvi:
        for pixel in row:
            if pixel > 1 or pixel < -1:
                print(fr'{filename} is no good!')
                bad_list.append(filename)
                return
    good_list.append(filename)

#Places good images into a folder within this notebook's directory
def save_good_files_ndvi(good_list, folder_directory):
    new_path = Path('Usable_NDVI_Images') #Can change folder name here
    new_path.mkdir(exist_ok=True)
    destination_path = new_path.resolve()
    for item in good_list:
        source_path = Path(fr'{folder_directory}/{item}')
        shutil.copy(source_path, destination_path)
```

```python
#Runs the script for NDVI tiles
bad_ndvi = []
good_ndvi = []
for item in os.listdir(ndvi_tile_folder_path):
    find_missing_data_ndvi(item, bad_ndvi, good_ndvi, ndvi_tile_folder_path)

save_good_files_ndvi(good_ndvi, ndvi_tile_folder_path)
```
